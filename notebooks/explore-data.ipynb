```python
def make_transit_request(url):
    headers ={
        # Request headers
        'Cache-Control': 'no-cache',
    }

    params = {
        'app_key': os.getenv('TUBE_KEY')
    }

    r = requests.get(url, headers=headers, params=params)

    return r.json()
```
